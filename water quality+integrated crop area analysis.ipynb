```python
df_clean = df.dropna(subset=['Type of water body'])

# Find the row with the highest temperature
max_temp_row = df_clean.loc[df_clean['Maximum temperature of water'].idxmax()]

# Print result
print("Water body with highest max temperature:", max_temp_row['Type of water body'])
print("Max temperature:", max_temp_row['Maximum temperature of water'])
```

```python
min_conductivity=df.loc[df['Minimum conductivity required for water quality'].idxmin()]
print("state with minimum conductivity of water quality: ",min_conductivity['srcStateName'])
print("min water conductivity",min_conductivity['Minimum conductivity required for water quality'])
```

```python
max_demand=df.loc[df['Maximum biochemical oxygen demand (BOD) required for water quality'].idxmax()]
print ("year with maximum BOD demand",max_demand['srcYear'])
print("max demand of bod is",max_demand['Maximum biochemical oxygen demand (BOD) required for water quality'] )
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Type of water body']=le.fit_transform(df['Type of water body'])
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['Type of water body'],df['Maximum temperature of water'])
plt.xlabel('Type of water body')
plt.ylabel('Mximum temperature of water')
plt.show()
```
